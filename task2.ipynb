```python
import pandas as pd
import numpy as np

# =========================================================
# 1. Load Dataset
# =========================================================

df = pd.read_csv("../Womens Clothing E-Commerce Reviews.csv")

print("Original Shape:", df.shape)


# =========================================================
# 2. Create Custom Cleaning Pipeline
# =========================================================


def clean_dataset(df):

    data = df.copy()
    cleaning_log = []

    # -----------------------------------------------------
    # Initial information
    # -----------------------------------------------------

    initial_rows = len(data)
    initial_columns = data.shape[1]

    cleaning_log.append(
        f"Initial dataset: {initial_rows} rows, {initial_columns} columns"
    )

    # -----------------------------------------------------
    # Detect Duplicate Rows
    # -----------------------------------------------------

    duplicate_count = data.duplicated().sum()

    if duplicate_count > 0:

        cleaning_log.append(f"Duplicate rows detected: {duplicate_count}")

        data = data.drop_duplicates().reset_index(drop=True)

        cleaning_log.append(f"Duplicate rows removed: {duplicate_count}")

    else:

        cleaning_log.append("No duplicate rows detected")

    # -----------------------------------------------------
    # Detect Irrelevant Columns
    # -----------------------------------------------------

    irrelevant_columns = []

    for column in data.columns:

        # Empty/unnamed columns
        if str(column).lower().startswith("unnamed"):

            irrelevant_columns.append(column)

        # Completely empty columns
        elif data[column].isna().all():

            irrelevant_columns.append(column)

        # Columns containing only one unique value
        elif data[column].nunique(dropna=False) <= 1:

            irrelevant_columns.append(column)

    # -----------------------------------------------------
    # Protect Important Columns
    # -----------------------------------------------------

    protected_columns = []

    for column in data.columns:

        column_name = str(column).lower()

        if any(
            keyword in column_name for keyword in ["id", "rating", "recommended", "age"]
        ):
            protected_columns.append(column)

    # Remove protected columns from irrelevant list

    irrelevant_columns = [
        column for column in irrelevant_columns if column not in protected_columns
    ]

    # -----------------------------------------------------
    # Log Detected Irrelevant Columns
    # -----------------------------------------------------

    if irrelevant_columns:

        cleaning_log.append(f"Irrelevant columns detected: {irrelevant_columns}")

    else:

        cleaning_log.append("No irrelevant columns detected")

    # -----------------------------------------------------
    # Remove Irrelevant Columns
    # -----------------------------------------------------

    if irrelevant_columns:

        data = data.drop(columns=irrelevant_columns)

        cleaning_log.append(f"Irrelevant columns removed: {irrelevant_columns}")

    # -----------------------------------------------------
    # Final Information
    # -----------------------------------------------------

    final_rows = len(data)
    final_columns = data.shape[1]

    cleaning_log.append(f"Final dataset: {final_rows} rows, {final_columns} columns")

    # -----------------------------------------------------
    # Display Cleaning Log
    # -----------------------------------------------------

    print("\nCleaning Log")
    print("-" * 50)

    for log in cleaning_log:
        print(log)

    # -----------------------------------------------------
    # Return Clean Dataset + Log
    # -----------------------------------------------------

    return data, cleaning_log


# =========================================================
# 3. Run Cleaning Pipeline
# =========================================================

df_clean, cleaning_log = clean_dataset(df)


# =========================================================
# 4. Verify Duplicate Removal
# =========================================================

print("\nRemaining duplicate rows:", df_clean.duplicated().sum())


# =========================================================
# 5. Compare Before and After
# =========================================================

print("\nBefore Cleaning:", df.shape)
print("After Cleaning :", df_clean.shape)


# =========================================================
# 6. View Clean Dataset
# =========================================================

df_clean.head()
```

```python
df_clean.to_csv("cleaned_womens_clothing.csv", index=False)
```
